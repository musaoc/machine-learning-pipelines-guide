## Machine Learning Pipelines Made Easy 🎹⛏🧺
The goals of this notebook is to make use of ML Pipelines simple for everyone.

___

### Why use Machine Learning Pipelines at the first place?
**Pipelines help you:**
1. Write a cleaner code
2. You don't have to do every step manually for train,val or test data 
3. It also helps you in deployment.

# 1.Necessary Libraries 😀

```python
import pandas as pd
import seaborn as sns
# Sklearn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,accuracy_score
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn import set_config
set_config(display='diagram')
```

# 2- Loading the Data 📅

```python
df_train = pd.read_csv('../input/titanic/train.csv')
df_test = pd.read_csv('../input/titanic/test.csv')

df_train.head()
```

# 3- Let's Explore 👓

```python
df_train.info()
```

```python
X=df_train.drop('Survived',axis=1)
y = df_train['Survived']
```

```python
X_train,X_val,y_train,y_val= train_test_split(X,y,test_size=0.25,random_state=1)
```

```python
cat_cols=X_train.select_dtypes('object').columns.to_list()
cat_cols
```

```python
num_cols=X_train.select_dtypes(exclude='object').columns.to_list()
num_cols
```

# 4- Creating Seperate Preprocessors for Numeric and Categorical Columns ♦

```python
# creating pipeline for numeric data preprocessing
numeric_preprocessor = Pipeline(steps=[
    ('imputer',SimpleImputer(strategy='mean')),
    ('scaling',StandardScaler()),
])
```

```python
# creating pipeline for categorical data preprocessing

categorical_preprocessor = Pipeline(steps=[
     ('imputer',SimpleImputer(strategy='constant')),
    ('encoder',OneHotEncoder(handle_unknown='ignore'))
])
```

# 5 - Using ColumnTransformer to transform categorical and numeric columns 😻

```python
preprocessor = ColumnTransformer([
    ('categorical',categorical_preprocessor,cat_cols),
    ('numeric',numeric_preprocessor,num_cols),

])
```

# 6 -  Its Pipeline Time 😎

```python
# adding our preprocessor and model in pipelines.
Pipe = Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('model',LogisticRegression(C=1e9)),
])
```

```python
Pipe.fit(X_train,y_train)
```

```python
pred=Pipe.predict(X_val)
pred_train = Pipe.predict(X_train)
print('Accuracy on training data :',accuracy_score(pred_train,y_train))
print('Accuracy on validation data :',accuracy_score(pred,y_val))
```

```python
sns.heatmap(confusion_matrix(pred,y_val),annot=True);
```

# 7. Prediction and Submission 😼🙌

```python
pred = Pipe.predict(df_test)

output = pd.DataFrame({'PassengerId': df_test.PassengerId,
                       'Survived': pred})
output.to_csv('submission.csv', index=False)
```
